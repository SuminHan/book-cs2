### Summary: class and object

- **Object** = data (state variables) + the functions that work on that data, in one bundle
- **Class** = the type of an object: the **cookie cutter** that makes objects

**Cookie cutter**: a class is a cookie cutter, an object is one cookie made with it.
- There is only one cutter (`Point`), but we can make as many cookies (`p1`, `p2`, ...) as we want.
- Every cookie has the same shape, but each one has its own values
  → **each object has its own state variables (`self.x`, `self.y`).**
- Unlike a cookie, an object can do things with its functions (`p1.distance(p2)`).

| | `Point` | `Person` (today's market) |
|---|---|---|
| class (cutter) | `class Point:` | `class Person:` |
| object (cookie) | `p1 = Point(1, 2)` | `kim = Person("Kim", 3000, [])` |
| state variables | `self.x`, `self.y` | `self.name`, `self.money`, `self.items` |
| functions | `distance`, `add`, `move`, ... | `buy`, `sell`, ... |

When we call `p1.distance(p2)`, `self` is `p1` and `p` is `p2`.

### Modifier and pure function

- **modifier**: changes the state variables of the object. Usually returns nothing (`None`).
  (e.g. `__init__`, `setX`, `move`)
- **pure function**: does not change any object. It returns a value, or a **new** object.
  (e.g. `__str__`, `distance`, `add`)

### Preparation: the class `Point`

Problems 1 and 3 use the class `Point` below. **Run this cell first.**
(When Colab restarts the runtime, the definitions are gone, so run this cell again.)

The comment above each function (`# modifier`, `# pure function`) tells the kind of the function.
Look at the output of `point_main()` to see the difference between `add` (pure) and `move` (modifier).

```python
class Point:
    # modifier
    def __init__(self, px, py):
        self.x = px
        self.y = py

    # pure function
    def __str__(self):
        return "(" + str(self.x) + "," + str(self.y) + ")"

    # pure function
    def getX(self):
        return self.x

    # pure function
    def getY(self):
        return self.y

    # modifier
    def setX(self, v):
        self.x = v

    # modifier
    def setY(self, v):
        self.y = v

    # pure function
    def distance(self, p):
        dx = self.x - p.x
        dy = self.y - p.y
        return (dx**2 + dy**2) ** 0.5

    # pure function
    def add(self, p):
        return Point(self.x + p.x, self.y + p.y)

    # modifier
    def move(self, dx, dy):
        self.x = self.x + dx
        self.y = self.y + dy


def point_main():
    p1 = Point(1, 2)
    p2 = Point(4, 6)
    print(p1, p2)               # (1,2) (4,6)
    print(p1.distance(p2))      # 5.0
    p3 = p1.add(p2)             # pure function: a NEW Point
    print(p1, p2, p3)           # (1,2) (4,6) (5,8)
    r = p1.move(10, 10)         # modifier: p1 itself changes
    print(p1, r)                # (11,12) None

point_main()
```
