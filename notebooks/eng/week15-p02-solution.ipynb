# Week 15 Problem 2: `Rational`

Given code (do not change):

```python
def gcd(a, b):
    a, b = abs(a), abs(b)
    while b != 0:
        a, b = b, a % b
    return a
```

Fill in the methods of `Rational`. **Run this cell again after every change.**

```python
class Rational:
    def __init__(self, n, d):
        # (1) n is the numerator, d is the denominator
        self.n = n
        self.d = d

    def __str__(self):
        # (2) return a string like "4/8"
        return str(self.n) + "/" + str(self.d)

    def invert(self):
        # (3) modifier: n/d -> d/n
        self.n, self.d = self.d, self.n

    def reduce(self):
        # (4) modifier: reduce to lowest terms using gcd (4/8 -> 1/2)
        g = gcd(self.n, self.d)
        self.n //= g
        self.d //= g

    def mul(self, r):
        # (5) pure: return a NEW, reduced Rational for self * r   (a/b * c/d = ac/bd)
        q = Rational(self.n * r.n, self.d * r.d)
        q.reduce()
        return q

    def add(self, r):
        # (6) pure: return a NEW, reduced Rational for self + r   (a/b + c/d = (ad+bc)/bd)
        q = Rational(self.n * r.d + r.n * self.d, self.d * r.d)
        q.reduce()
        return q
```

**(1) `__init__`**: n is the numerator, d is the denominator

```python
a = Rational(4, 8)
print(a.n, a.d)          # 4 8
```

**(2) `__str__`**: return a string like "4/8"

```python
a = Rational(4, 8)
print(a)                 # 4/8
```

**(3) `invert`**: modifier: n/d -> d/n

```python
b = Rational(1, 3)
b.invert()
print(b)                 # 3/1
```

**(4) `reduce`**: modifier: reduce to lowest terms using gcd (4/8 -> 1/2)

```python
a = Rational(4, 8)
a.reduce()
print(a)                 # 1/2
```

**(5) `mul`**: pure: return a NEW, reduced Rational for self * r   (a/b * c/d = ac/bd)

```python
a = Rational(1, 2)
b = Rational(2, 3)
print(a.mul(b))          # 1/3
print(a, b)              # 1/2 2/3
```

**(6) `add`**: pure: return a NEW, reduced Rational for self + r   (a/b + c/d = (ad+bc)/bd)

```python
a = Rational(1, 2)
b = Rational(1, 6)
print(a.add(b))          # 2/3
print(a, b)              # 1/2 1/6
```
