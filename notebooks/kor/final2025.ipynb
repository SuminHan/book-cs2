이렇게 생성한 임의의 리스트 `M`에 대해 `P15(M)`의 최대값은 무엇인가?
(제시한 최대값이 $S$라고 하면, `P15(M*)`가 $S$가 되는 `M*`을 하나 구성하고 `P15(M)` $> S$인 `M`이 존재하지 않음을 증명해야 한다.)

`P15` 함수는 아래 셀에 있습니다. 시험장에서는 실행해 볼 수 없으니 **먼저 종이에서 생각해 본 다음** 실행해 보세요.

````python
class Card:
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def lessThan(self, card):
        if self.rank != card.rank:
            return self.rank < card.rank
        else:
            return self.suit < card.suit

def sortRow(M, r):
    h = len(M)
    w = len(M[0])
    R = [[None]*w for i in range(h)]
    for i in range(h):
        for j in range(w):
            R[i][j] = M[i][j]
    for j in range(w-1):
        for k in range(j+1,w):
            if R[r][k].lessThan(R[r][j]):
                R[r][j], R[r][k] = R[r][k], R[r][j]
    return R

def sortCol(M, c):
    h = len(M)
    w = len(M[0])
    R = [[None]*w for i in range(h)]
    for i in range(h):
        for j in range(w):
            R[i][j] = M[i][j]
    for i in range(h-1):
        for k in range(i+1,h):
            if R[k][c].lessThan(R[i][c]):
                R[i][c], R[k][c] = R[k][c], R[i][c]
    return R

def countInversions(M):
    h = len(M)
    w = len(M[0])
    counter = 0
    for i in range(h):
        for j in range(w-1):
            for k in range(j+1,w):
                if M[i][k].lessThan(M[i][j]):
                    counter += 1
    return counter

def P15(M):
    for i in range(4):
        M = sortRow(M, i)
    for j in range(13):
        M = sortCol(M, j)
    return countInversions(M)

# ===== 실험용: 무작위로 섞은 M에 대해 P15(M)을 여러 번 계산 =====
def shuffled(seed):
    rng = random.Random(seed)
    M = [[None]*13 for i in range(4)]
    for i in range(4):
        for j in range(13):
            M[i][j] = Card(i, j+1)
    for k in range(100000):
        i1 = rng.randrange(4)
        j1 = rng.randrange(13)
        i2 = rng.randrange(4)
        j2 = rng.randrange(13)
        M[i1][j1], M[i2][j2] = M[i2][j2], M[i1][j1]
    return M

for seed in range(5):
    print(P15(shuffled(seed)))
````
