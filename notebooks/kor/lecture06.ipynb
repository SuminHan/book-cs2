import random

    for h in range(len(A)):
        f[A[h][0]] = A[h][1]

    T = 1000000  # sufficiently large number
    total_trials = 0
    for k in range(T):
        pos = 0
        num_trials = 0
        while True:
            num_trials += 1
            step = random.randrange(1, m+1)
            pos = f[pos + step]
            if pos >= n:
                break
        total_trials += num_trials
    return total_trials / float(T)
